This Code enables easy control of the Bigfoot Software through Python. The Bigfoot executable needs to be running and connected to **all components** (MONSTR Sense electronics & Stages) for this Python library to control the executable. The following controls/readouts are accessible through the Python library:

- Move stages to specified position
- Change spectral parameters
- Change scan parameters
- Change basic scan parameters
- Run scan
- Stop scan
- Check stage movement
- Check whether a scan/real-time is running
- Read Effective Wavelength & suggested step sizes

Please keep in mind that as you construct automation from this library, checks on whether stages are moving or scans are running are important to operate a smooth and issue-free operation. The Bigfoot executable has many checks built-in to ensure that a command is not executed if it conflicts with the current operating state of the system, e.g., moving stages during a scan. In this case, the executable will return a "Scan in Progress" error to Python. .

To run this code, you need to install the JKI Python Bridge: *pip install jki-python-bridge-for-labview*

## Import libraries & custom functions

```python
from jki_python_bridge_for_labview import labview as lv


unit_map = {
    'nm': 0,
    'meV': 1,
    'THz': 2,
    'cm-1': 3
}

def units(string):
    try:
        return unit_map[string]
    except KeyError:
        raise ValueError(f"Unknown unit: {string}")
```

## Establish server connection 

```python
lv.connect()
serv_connect = lv.isConnected
print(serv_connect)
```

## Python-Labview Comms Functions

#### Stage movement

```python
#move_stage_pos(stage_sel,position)
#stage_sel: 0(tau), 1 (T), 2 (t)
#position: in ps

#Example:
lv.LV_Control.move_stage_pos(2,5)
```

#### Change spectral parameters

```python
#change_spectrum(center, unit1, bandwidth, unit2)
#center & unit1: Signal Spectreal center and corresponding unit
#bandwidth & unit2: Signal bandwidth and corresponding unit

#Example:
lv.LV_Control.change_spectrum(840,units('nm'),250,units('nm'))
```

#### Change Scan parameters

```python
#change_scan(type, t_length, step_length, fixed_delay)
#type: 1Q (Single Quantum), 0Q (Zero Quantum), 2Q (Double Quantum), 1Q-NR (Rephasing), 1Q-NR (Non-rephasing)
#t_length: t-axis Scan Length in ps
#step_length: 2D-axis Scan Length in ps
#fixed_delay: Fixed Delay (pulse width) in ps

#Example:
lv.LV_Control.change_scan('1Q-R',0.12,0.12,1)
```

#### Change Basic Scan parameters

```python
#change_basic_scan(min_width,neg_start,oversample)
#min_width: Scan Min. Width in ps
#neg_start: 2D-axis Neg. Start in ps
#oversample: Oversample 2D

#Example:
lv.LV_Control.change_basic_scan(0.8,0.2,1)
```

#### Run Scan

```python
#Starts an MDCS Scan with the currently set parameters

#Example:
lv.LV_Control.run_scan()
```

#### Stop Scan 

```python
#Stops the MDCS Scan

#Example:
lv.LV_Control.stop_scan()
```

#### Acquire Phase

```python
#Acquires a single datapoint of the Ref A/B Phase and Ref C/D Phase

#Example:
phase_AB, phase_CD = lv.LV_Control.acquire_phase()
print(phase_AB,phase_CD)
```

#### Check Stage Movement

```python
#Returns array of stage movement true (1)/false (0) in the following order: tau/T/t

#Example:
lv.LV_Control.check_stage_move()
```

#### Check Scan/Real-time

```python
#Returns array of whether real-time is running (first boolean) or scan is in progress (second boolean)

#Example:
lv.LV_Control.check_scan()
```

#### Read effective wavelength & suggested step size

```python
#Returns three values: effective wavelength (nm), suggested step size (t-axis), suggested oversamplesd step size (2D-axis)

#Example:
lv.LV_Control.read_scan_params()
```
